```python
required_cols = ["Age", "speech", "python", "math", "english"]
subjects = ["python", "math", "english", "speech"]

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["speech"] = pd.to_numeric(df["speech"], errors="coerce")

df4 = df.dropna(subset=required_cols).copy()

df4

df4["Total"] = df4[subjects].sum(axis=1)
df4["Average"] = df4[subjects].mean(axis=1)
df4["Subjects_Attempted"] = df4[subjects].notna().sum(axis=1)

df4
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df4["Student_Name"], df4["Total"])
plt.xticks(rotation=90)
plt.xlabel("Student")
plt.ylabel("Total Marks")
plt.title("Total Marks per Student")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["python", "math", "english", "speech"]
subject_avg = df4[subjects].mean()

plt.figure()
plt.plot(subject_avg.index, subject_avg.values, marker='o')
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.title("Average Score per Subject")
plt.grid(True)
plt.show()
```
